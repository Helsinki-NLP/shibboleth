```python
import collections
import itertools
import json
import pandas as pd
import os
```

```python
FILES = [
    ('scandinavian', '../../data/scandinavian/slide_sl_all_tok.csv'),
    ('estonian_voro', '../../data/estonian_voro/pkev_all_tok.csv'),
    ('bcms_setimes', '../../data/cs_setimes/all_tok.tsv'),
]
```

```python
os.makedirs('freqs', exist_ok=True)

for grp, file in FILES:
    data = pd.read_csv(file, header=None, names=['label', 'raw', 'tok'], sep='\t')
    for label, subdf in data.groupby('label'):
        freqs = dict(collections.Counter(itertools.chain.from_iterable(subdf.dropna().tok.str.split().to_list())))
        with open(f'freqs/freq.{grp}.{label}.json', 'w') as ostr:
            json.dump(freqs, ostr, ensure_ascii=False)
```
