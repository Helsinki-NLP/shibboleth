```python
import seaborn as sns
import pandas as pd
import altair as alt
```

```python
df = pd.read_csv('auroc_data+wl.csv').sort_values(['lang group', 'attrib'])
df['Attribution'] =  df['attrib'].apply(lambda fname: fname.split('_')[0].upper())
df['Group'] = df['lang group'].map({
    'estonian_voro': 'EV', 
    'bcms_twitter': 'BCMS (T)', 
    'bcms_setimes': 'BCMS (S)',
    'bcms': 'BCMS (S)',
    'middle_low_saxon': 'MLS',
    'finnish': 'Finnish',
    'greek': 'Greek',
    'jodel': 'German',
    'scandinavian': 'Scandi.',
})

EXP_1_SET = {'Scandi.', 'BCMS (S)', 'EV'}
df
```

```python
g = sns.stripplot(data=df, x='Group', y='AUROC', hue='Attribution', dodge=True)
g.set(ylim=(0,100))
sns.move_legend(g, "upper left", bbox_to_anchor=(1, 1))
g.get_figure().savefig('AUROCS_+wl.pdf', bbox_inches='tight')
```

```python
df = pd.read_csv('evals_test_all.csv', header=None, names=['Group', 'iter', 'fold', 'Accuracy'])
df['Group'] = df['Group'].map({
    'estonian_voro': 'EV', 
    'bcms_twitter': 'BCMS (T)', 
    'bcms_setimes': 'BCMS (S)',
    'middle_low_saxon': 'MLS',
    'finnish': 'Finnish',
    'greek': 'Greek',
    'jodel': 'German',
    'scandinavian': 'Scandi.',
})
df
```

```python
g = sns.stripplot(data=df, x='Group', y='Accuracy')
g.set(ylim=(0,1))
g.get_figure().savefig('acc-heldout.pdf', bbox_inches='tight')
```

```python
g = sns.stripplot(data=df[df.Group.isin(EXP_1_SET)], x='Group', y='Accuracy')
g.set(ylim=(0,1))
g.get_figure().savefig('acc-heldout.exp1.pdf', bbox_inches='tight')
```

```python
df[df.Group.isin(EXP_1_SET)].groupby('Group').Accuracy.min(), df[df.Group.isin(EXP_1_SET)].groupby('Group').Accuracy.max()
```

```python
df_ = df.merge((df.groupby('Group')['Accuracy'].mean() - 0.05).reset_index().rename(columns={'Accuracy': 'threshold'}))
df_[df_['Accuracy'] < df_['threshold']]
```
