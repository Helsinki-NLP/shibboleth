```python
import seaborn as sns
import pandas as pd
import altair as alt


sns.set(font_scale=2) 
sns.set_style("whitegrid")

EXP_1_SET = {'Scandi.', 'CS', 'EV'}
EXP_2_SET = {'BCMS', 'German', 'Greek', 'Finnish'}
```

```python
df = pd.read_csv('../../analyses/auroc_data+wl.csv').sort_values(['lang group', 'attrib'])
df['Attribution'] =  df['attrib'].apply(lambda fname: fname.split('_')[0].upper())
df['Group'] = df['lang group'].map({
    'estonian_voro': 'EV', 
    'bcms_twitter': 'BCMS', 
    'bcms_setimes': 'CS',
    'bcms': 'CS',
    'middle_low_saxon': 'MLS',
    'finnish': 'Finnish',
    'greek': 'Greek',
    'jodel': 'German',
    'scandinavian': 'Scandi.',
})

df
```

```python
g = sns.stripplot(data=df, x='Group', y='AUROC', hue='Attribution', hue_order=sorted(df.Attribution.unique()), dodge=True)
g.set(ylim=(0,100))
sns.move_legend(g, "upper left", bbox_to_anchor=(1, 1))
g.get_figure().savefig('AUROCS_+wl.pdf', bbox_inches='tight')
```

```python
df = pd.read_csv('../../analyses/evals_test_all.csv', header=None, names=['Group', 'iter', 'fold', 'Accuracy'])
df['Group'] = df['Group'].map({
    'estonian_voro': 'EV', 
    'bcms_twitter': 'BCMS', 
    'bcms_setimes': 'CS',
    'middle_low_saxon': 'MLS',
    'finnish': 'Finnish',
    'greek': 'Greek',
    'jodel': 'German',
    'scandinavian': 'Scandi.',
})
df
```

```python
df_ = df[df.Group.isin(EXP_1_SET)].sort_values('Group')
g = sns.stripplot(data=df_, x='Group', y='Accuracy', hue='Group', hue_order=sorted(df_.Group.unique()))
g.set(ylim=(0,1))
g.get_figure().savefig('acc-heldout.exp1.pdf', bbox_inches='tight')
```

```python
df_ = df[df.Group.isin(EXP_2_SET)].sort_values('Group')

g = sns.stripplot(data=df_, x='Group', y='Accuracy', hue='Group', hue_order=sorted(df_.Group.unique()))
g.set(ylim=(0,1))
g.get_figure().savefig('acc-heldout.exp2.pdf', bbox_inches='tight')
```

```python
df[df.Group.isin(EXP_1_SET)].groupby('Group').Accuracy.min(), df[df.Group.isin(EXP_1_SET)].groupby('Group').Accuracy.max()
```

```python
df_ = df.merge((df.groupby('Group')['Accuracy'].mean() - 0.05).reset_index().rename(columns={'Accuracy': 'threshold'}))
df_[df_['Accuracy'] < df_['threshold']]
```
